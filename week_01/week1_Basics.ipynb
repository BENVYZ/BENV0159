<font size="5">Key features of dataframe</font><br>
- **Two-Dimensional Structure**: A DataFrame is inherently two-dimensional, meaning it has rows and columns. It can be thought of as a table or a spreadsheet where data is organized in a tabular format.<br>
<br>
- **Labeled Axes (Rows and Columns)**
   - Row Labels (Index):Each row in a DataFrame is labeled with an index. By default, this index is an integer ranging from 0 to n-1, where n is the number of rows. However, you can customize the index to be any set of labels, such as strings, dates, or other data types.
   - Column Labels:Each column in a DataFrame is also labeled, typically with a string that describes the data in that column. These labels make it easy to access and manipulate data by referring to column names rather than numeric positions.<br>   
<br>   
- **Heterogeneous Data Types**: A DataFrame can contain columns with different data types. For example, one column can store integers, another can store floating-point numbers, and another can store strings. This flexibility makes DataFrames ideal for real-world datasets that often contain mixed types of data.

<br>

- values: The values are the actual data stored in the DataFrame, represented as a 2D array. These are the entries or cells in the table, where each value is associated with a specific row and column.

#### df.iloc: selection by position

To help you remember, iloc stands for "integer location."

Note: the row names (labels) and row integer locations in a DataFrame might coincide, especially when the row labels are simple integers starting from 0.

```python
df_house.iloc[ 1:3 ,  : ]

#The comma separates the row selection from the column selection.
#Since no column selection is specified after the comma, it implies that all columns should be selected.
#Important Note: in iloc slicing, the start index is inclusive, but the stop index is exclusive. 
#e.g., for this case, we start at 1, go up to but do not include 3
```

```python
df_house.loc[ 1:3 , :  ]
```

! important: When using df.loc[1:3, ], the input range 1 to 3 includes rows labeled 1 to 3, unlike df.iloc[1:3, ], which only includes rows from 1 to 2.

#### slice by boolean
A Boolean in Python is a data type that represents one of two possible values: True or False. 


```python
# now i only want to check which boroughs worth investment

df_house[df_house['Investment'] == 'yes']
```

```python
(df_house['Investment'] == 'yes' ) & (df_house['MedianHousePrice']< 260000 )
```

```python
#Now I would like to refine my selection criteria: 
#not only should it be a worthwhile investment, 
#but the median house price must also be lower than 26000.


df_house [ (df_house['Investment'] == 'yes' ) & (df_house['MedianHousePrice']< 260000 ) ]
```

```python
# We now use the | operator to filter rows where either the investment is worthwhile
# or the median price is less than 260,000. 
#The output will include rows that meet at least one of these conditions.

df_house [ (df_house['Investment'] == 'yes' ) | (df_house['MedianHousePrice']< 260000 ) ]
```
